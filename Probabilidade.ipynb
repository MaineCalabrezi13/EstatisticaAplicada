```python
# ============================================
# CÉLULA 1 — Probabilidade clássica: dado
# ============================================

# Espaço amostral: todos os resultados possíveis
espaco_amostral = [1,2,3,4,5,6]       # complete aqui

# Evento: resultados que nos interessam
evento_par = [2,4,6]            # complete aqui

n_omega  = len(espaco_amostral)
n_evento = len(evento_par)
prob     = n_evento / n_omega

print(f"P(par) = {n_evento}/{n_omega} = {prob:.2f} = {prob*100:.0f}%")
```

```python
# ============================================
# CÉLULA 2 — Urna com 8 bolas
# ============================================
espaco_amostral = [1, 2, 3, 4, 5, 6, 7, 8]
n_omega = len(espaco_amostral)

evento_par    = [2, 4, 6, 8]   # já pronto — exemplo
evento_impar  = [1,3,5,7]           # complete
evento_menor2 = [1]           # complete
evento_entre  = [5,6,7]           # complete

prob_par    = len(evento_par)    / n_omega
prob_impar  = len(evento_impar)  / n_omega
prob_menor2 = len(evento_menor2) / n_omega
prob_entre  = len(evento_entre)  / n_omega

print(f"a) P(par)         = {prob_par*100:.1f}%")
print(f"b) P(ímpar)       = {prob_impar*100:.1f}%")
print(f"c) P(menor que 2) = {prob_menor2*100:.1f}%")
print(f"d) P(entre 4 e 8) = {prob_entre*100:.1f}%")
print(f"Soma par+ímpar    = {prob_par+prob_impar}")
```
