```python
# ============================================
# CÉLULA 0 — Importação de bibliotecas
# Execute esta célula PRIMEIRO
# ============================================

import random          # sorteios aleatórios
import matplotlib.pyplot as plt  # gráficos

print("Bibliotecas carregadas com sucesso!")
```

```python
# ============================================
# CÉLULA 1 — Probabilidade clássica: dado
# ============================================

# Espaço amostral: todos os resultados possíveis
espaco_amostral = [1,2,3,4,5,6]       # complete aqui

# Evento: resultados que nos interessam
evento_par = [2,4,6]            # complete aqui

n_omega  = len(espaco_amostral)
n_evento = len(evento_par)
prob     = n_evento / n_omega

print(f"P(par) = {n_evento}/{n_omega} = {prob:.2f} = {prob*100:.0f}%")
```

```python
# ============================================
# CÉLULA 2 — Urna com 8 bolas
# ============================================
espaco_amostral = [1, 2, 3, 4, 5, 6, 7, 8]
n_omega = len(espaco_amostral)

evento_par    = [2, 4, 6, 8]   # já pronto — exemplo
evento_impar  = [1,3,5,7]           # complete
evento_menor2 = [1]           # complete
evento_entre  = [5,6,7]           # complete

prob_par    = len(evento_par)    / n_omega
prob_impar  = len(evento_impar)  / n_omega
prob_menor2 = len(evento_menor2) / n_omega
prob_entre  = len(evento_entre)  / n_omega

print(f"a) P(par)         = {prob_par*100:.1f}%")
print(f"b) P(ímpar)       = {prob_impar*100:.1f}%")
print(f"c) P(menor que 2) = {prob_menor2*100:.1f}%")
print(f"d) P(entre 4 e 8) = {prob_entre*100:.1f}%")
print(f"Soma par+ímpar    = {prob_par+prob_impar}")
```

```python
# ============================================
# CÉLULA 3 — Evento complementar
# P(Ā) = 1 - P(A)
# ============================================

p_menina = 0.56

# Complete: calcule o complementar
p_menino = 1 - p_menina

print(f"P(menina) = {p_menina}")
print(f"P(menino) = {p_menino}")
print(f"Soma total = {round(p_menina + p_menino, 2)}")
print(f"Probabilidade de menino: {p_menino*100:.0f}%")
```

```python
# ============================================
# CÉLULA 4 — Probabilidade frequencial
# ============================================

total_lancamentos = 100   # complete: use 100
contagem_par = 0

for i in range(total_lancamentos):
    resultado = random.randint(1, 6)
    if resultado % 2 == 0:               # complete: condição de par
        contagem_par += 1

prob = contagem_par / total_lancamentos
print(f"Pares: {contagem_par} de {total_lancamentos}")
print(f"Probabilidade obtida:   {prob:.2f} = {prob*100:.1f}%")
print(f"Probabilidade clássica: 0.50 = 50%")
print(f"Diferença: {abs(prob - 0.5):.3f}")
```

```python
# ============================================
# CÉLULA 5 — Gráfico de barras
# ============================================

total = 60   # complete: use 60
resultados = [random.randint(1, 6) for _ in range(total)]

faces       = [1, 2, 3, 4, 5, 6]
frequencias = [resultados.count(f) for f in faces]

plt.figure(figsize=(8, 4))
plt.bar(faces, frequencias, color="steelblue", edgecolor="white")
plt.axhline(y=total/6, color="red", linestyle="--", label="Esperado")
plt.title("Frequência das faces do dado")    # complete: título do gráfico
plt.xlabel("Face do dado")   # complete: nome do eixo X
plt.ylabel("Frequência")
plt.xticks(faces)
plt.legend()
plt.show()
```
